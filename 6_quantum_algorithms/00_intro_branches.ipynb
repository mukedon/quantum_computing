# Categorization of Quantum Algorithms

**Sources:**

1. [Quantum algorithms:A survey of applications and end-to-end complexities](https://arxiv.org/pdf/2310.03011)

Quantum Algorithms (QAs) are categorized into major branches based on the underlying mathematical framework, problem domain, and hardware execution paradigm.

## 1. Foundational & Oracle Algorithms

These algorithms rely on [black-box functions (oracles)](https://en.wikipedia.org/wiki/Black_box) to evaluate global properties of functions faster than classical methods. They established the theoretical foundation of quantum speedup.

* [**Deutsch-Jozsa Algorithm:**](https://quantum.cloud.ibm.com/learning/en/courses/fundamentals-of-quantum-algorithms/quantum-query-algorithms/deutsch-jozsa-algorithm) Determines whether a black-box [Boolean function](https://en.wikipedia.org/wiki/Boolean_function) is constant of balanced in a single query.

* [**Bernstein-Vazirani Algorithm:**](https://en.wikipedia.org/wiki/Bernstein%E2%80%93Vazirani_algorithm) Finds a hidden bit string encoded within a black-box function in $\mathcal{O} (1)$ queries compared to $\mathcal{O}(n)$ classically.

* [**Simon's Algorithm:**](https://en.wikipedia.org/wiki/Simon%27s_problem) Identifies a hidden period in a function using XOR operations; inspired the period-finding technique behind Shor's algorithm.

* [**Hidden Subgroup Problem (HSP):**](https://en.wikipedia.org/wiki/Hidden_subgroup_problem) A general theoretical framework underlying many algebraic algorithms (including Shor's) across finite abelian and non-abelian groups.

## 2. Quantum Fourier Transform & Algebraic Algorithms

These leverage the [Quantum Fourier Transform (QFT)](https://en.wikipedia.org/wiki/Quantum_Fourier_transform) to solve period-finding, factoring and algebraic tasks with exponential speedup over classical algorithms.

* [**Shor's Algorithm:**](https://en.wikipedia.org/wiki/Shor%27s_algorithm) Factors integers and computes discrete logarithms in polynomical time ([time complexity](https://en.wikipedia.org/wiki/Time_complexity)), posing a direct threat to public-key [cryptography (RSA, ECC)](https://en.wikipedia.org/wiki/Cryptography).

* [**Quantum Phase Estimation (QPE):**](https://en.wikipedia.org/wiki/Quantum_phase_estimation_algorithm) Estimates the phase/eigenvalues of a unitary operator; serves as a key subroutine in chemistry and linear algebra algorithms.

* [**Hadamard Test**](https://en.wikipedia.org/wiki/Hadamard_test) & [**Swap Test:**](https://en.wikipedia.org/wiki/Swap_test) Basic circuit subroutines used to estimate expectation values or measure the overlap (similarity) between two quantum states.

* [**Aharonov-Jones-Landau Algorithm:**](https://en.wikipedia.org/wiki/Aharonov%E2%80%93Jones%E2%80%93Landau_algorithm) Approximates topological invariants (such as the Jones polynomial) using algebraic techniques.

## 3. Search & Amplitude Amplification Algorithms

Algorithms in this branch enhance the probability of target computational states, providing polynomial (typically quadratic) speedups for unstructured problems.

* [**Grover's Algorithm:**](https://en.wikipedia.org/wiki/Grover%27s_algorithm) Searches an unstructured database of $N$ items in $\mathcal{O}(\sqrt{N})$ time instead of $\mathcal{O}(N)$.

* [**Amplitude Amplification:**](https://en.wikipedia.org/wiki/Amplitude_amplification) A generalized extension of Grover's technique used across various algorithms to boost the probability of correct outcomes.

* [**Quantum Counting Algorithm:**](https://en.wikipedia.org/wiki/Quantum_counting_algorithm) Estimates the number of marked or valid solutions in unsorted search space.

* [**BHT Algorithm (Brassard-Heller-Tapp):**](https://en.wikipedia.org/wiki/BHT_algorithm) Solves the collision problem using a combination of quantum search and birthday attack concepts.

## 4. Quantum Simulation & Linear Algebra

Designed to solve high-dimensional matrix tasks, simulate quantum physics directly, and process systems of linear equations.

* [**Hamiltonian Simulation (Trotterization, QSP):**](https://en.wikipedia.org/wiki/Hamiltonian_simulation) Simulates the time evolution of physical systems governed by a quantum Hamiltonian, crucial for materials science and chemistry.

* [**HHL Algorithm (Harrow-Hassidim-Lloyd):**](https://en.wikipedia.org/wiki/HHL_algorithm) Solves large, sparse systems of linear equations $(Ax = b)$ with exponential speedup in system size under specific matrix constraints.

* [**Quantum Singular Value Transformation (QSVT):**](https://en.wikipedia.org/wiki/Quantum_singular_value_transformation) A unified linear algebra framework that generalizes matrix transformations, QPE, and amplitude amplification.

## 5. Optimization & Variational (Hybrid) Algorithms

Algorithms geared toward current Noisy Intermediate-Scale Quantum (NISQ) devices, combining classical optimization loops with parameterized quantum circuits.

* [**Variational Quantum Eigensolver (VQE):**](https://en.wikipedia.org/wiki/Variational_quantum_eigensolver) Estimates the ground-state energy of molecules and Hamiltonians by optimizing parameters classically.

* [**Quantum Approximate Optimization Algorithm (QAOA):**](https://en.wikipedia.org/wiki/Quantum_optimization_algorithms) Solves combinatorial and graph optimization problems (such as Max-Cut) using alternating cost and mixer operations.

* [**Quantum Annealing:**](https://en.wikipedia.org/wiki/Quantum_annealing) Solves Ising spin glass models and binary optimization problems via adiabatic evolution (e.g., [D-Wave system paradigms](https://en.wikipedia.org/wiki/D-Wave_Systems)).

## 6. Quantum Machine Learning & Sampling

Algorithms focused on statistical physics, data processing, and demonstrating quantum advantage over classical distribution sampling.

* [**Quantum Support Vector Machines (QSVM):**](https://arxiv.org/pdf/1307.0471) & [**Quantum Neural Networks (QNN):**](https://en.wikipedia.org/wiki/Quantum_neural_network) Model high-dimensional feature spaces using quantum kernel methods or parameterized quantum circuits.

* [**Boson Sampling:**](https://en.wikipedia.org/wiki/Boson_sampling) A non-universal optical model designed to prove quantum advantage by sampling photon interference patterns (exponentially hard to simulate classically).

* [**Quantum Random Walks (Discrete & Continuous):**](https://en.wikipedia.org/wiki/Quantum_walk) The quantum analogue of classical random walks, offering speedups for graph traversal, spatial search, network analysis.

## Detailed Comparison: Quantum vs. Classical Paradigms

* **Computational Mechanism:** Classical algorithms operate deterministically or stochastically on binary bits ($0$ or $1$) using Boolean logic gates (AND, OR, NOT). Quantum algorithms execute unitary operations on quantum bits (qubits) in superposition, utilizing constructive interference to amplify the probability of reaching the correct state and destructive interference to cancel out incorrect paths.

* **Problem Domains & Advantage Types:**
    * **Exponential Speedups:** Achieved primarily in algebraic, period-finding, and quantum simulation domains (e.g., Shor’s, QPE, Hamiltonian Simulation). These rely on global properties and the Quantum Fourier Transform.
    
    * **Quadratic Speedups:** Found in unstructured search and sampling problems (e.g., Grover’s, Quantum Counting). Quadratic speedups are provably optimal for black-box search and do not turn NP-complete problems into polynomial-time problems.

* **Limitations:** Quantum algorithms require reversible gate operations (unlike classical irreversible logic) and are vulnerable to environmental noise, requiring overhead for Quantum Error Correction (QEC) in fault-tolerant execution.Hybrid Quantum-Classical Computing BranchIn the Noisy Intermediate-Scale Quantum (NISQ) era, hardware lacks full error correction and is limited by restricted qubit counts and short coherence times. Hybrid Quantum-Classical Algorithms mitigate hardware noise by offloading parameter optimization to classical processing units while executing shallow, parameterized quantum circuits on quantum hardware.
