## Pauli Gates

### 1. Introduction
* The Pauli gates are the most basic single-qubit quantum gates.
* They correspond to rotations by $\pi$ radians $(180^\circ)$ around the three axes of Bloch sphere.
* These gates form the foundation for most quantum algorithms.

| Gate | Axis | Rotation angle | Matrix | Symbol |
|------|------|----------------|--------|--------|
| X | x-axis | $\pi$ | $\begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$ | NOT gate |
| Y | y-axis | $\pi$ | $\begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}$ | |
| Z | z-axis | $\pi$ | $\begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$ | Phase flip |

### 2. Geometric action on the Bloch Sphere

| Gate | Effect on $\|0\rangle$ | Effect on $\|1\rangle$ | Rotation |
|------|-------------|-------------|------------|
| X | flips $\|0\rangle \leftrightarrow \|1\rangle$ | Rotation about x-axis by $\pi$ |


Visual tip:

On the Bloch sphere:
* $X$ gate = half-turn around x-axis
* $Y$ gate = half-turn around y-axis
* $Z$ gate = half-turn around z-axis

### 3. Creating a quantum circuit with Pauli gates:

```python
# Imports
from qiskit import QuantumCircuit
from qiskit import QuantumRegister, QuantumCircuit
from qiskit.visualization import visualize_transition
import numpy as np
from numpy import pi

print('Qiskit version:', __import__('qiskit').__version__)
print("Environment ready.")
```

```python
# Create a quantum circuit
qc = QuantumCircuit(1)
qc.x(0)
qc.y(0)
qc.z(0)

# Plot the circuit
qc.draw('mpl', style='iqp-dark', fold=False)
```

One can visualize their action using the Bloch transition tool:

```python
# Register a qubit
qr = QuantumRegister(1)
qc = QuantumCircuit(qr)

# Apply each Pauli gate
qc.x(0)
qc.y(0)
qc.z(0)

# Animate vector transitions on Bloch sphere
visualize_transition(qc, fpg=20, spg=1)
```

### 4. Commutation and Anti-commutation
* Pauli matrices anticommute:
    * $\{ X, Y \} = \{ Y, Z \} = \{ Z, X \} = 0$
* They also square to identity:
    * $X^2 = Y^2 = Z^2 = I$
* This structure defines the Pauli group, a key mathematical backbone in quantum error correction and stabilizer theory.

### 5. Connection to Rotations
Each Pauli gate can be seen as a special case of a rotation:
$$
    X = R_x (\pi), \quad Y = R_y (\pi), \quad Z = R_z (\pi)
$$
Hence, the Pauli gates are the simplest nontrivial rotations in quantum mechanics.

### 6. Summary
* Pauli gates = π rotations around x, y, z axes.
* They are unitary, Hermitian, and involutory.
* Serve as the building blocks of Clifford circuits, measurements, and error correction codes.
